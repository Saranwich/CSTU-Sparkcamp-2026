**3. Logic, Control Flow & Filtering & Loop**

Logic, Control Flow, Filtering และ Loop คือ องค์ประกอบหลักที่ใช้ในการควบคุมทิศทางและการประมวลผลข้อมูลในโลกของการเขียนโปรแกรมเปรียบเสมือนสมองและระบบคิดของโค้ดที่ช่วยเปลี่ยนข้อมูลดิบให้กลายเป็นผลลัพธ์ตามเงื่อนไขที่ต้องการ

**1. Logic (ตรรกศาสตร์)จุดเริ่มต้นของการตัดสินใจในโปรแกรม โดยใช้ค่าความจริง True (จริง) หรือ False (เท็จ) มาคำนวณผ่านตัวดำเนินการ (Operators) เพื่อสร้างเงื่อนไข**

*   Comparison Operators: การเปรียบเทียบค่า เช่น == (เท่ากับ), != (ไม่เท่ากับ), >, <, >=, <=
*   Logical Operators: การเชื่อมตรรกะ เช่น AND (ต้องจริงทั้งคู่ถึงจะจริง) OR (จริงอย่างใดอย่างหนึ่งก็ถือว่าจริง) NOT (กลับค่าความจริงจากจริงเป็นเท็จ จากเท็จเป็นจริง)

**2. Control Flow (การควบคุมทิศทาง)กลไกที่ใช้กำหนดว่าโปรแกรมจะเลือกเดินไปที่เส้นทางไหน โดยประเมินผลมาจากเงื่อนไขของ Logic ด้านบน**

*   If-Else Statement:
คำสั่งพื้นฐานที่สุดในการเลือกทางเดินถ้าเงื่อนไขเป็นจริง ทำงานที่บล็อก ifถ้าเงื่อนไขเป็นเท็จ ข้ามไปทำงานที่บล็อก else
*  Switch / Case: ใช้ในกรณีที่มีตัวเลือกจำนวนมาก เพื่อหลีกเลี่ยงการเขียน if-else ซ้อนกันหลายชั้น

**3. Loop (การทำงานวนซ้ำ)คำสั่งที่สั่งให้โปรแกรมทำงานชุดเดิมซ้ำๆ เพื่อจัดการกับข้อมูลจำนวนมาก หรือทำงานอัตโนมัติจนกว่าจะบรรลุเป้าหมาย**


*   For Loop: ใช้เมื่อ รู้จำนวนครั้งที่แน่นอน หรือต้องการไล่ดูข้อมูลทีละตัวในกลุ่มข้อมูล (Array/List)
*   While Loop: ใช้เมื่อ ไม่รู้จำนวนครั้งที่แน่นอน แต่จะทำซ้ำไปเรื่อยๆ ตราบใดที่เงื่อนไขยังคงเป็นจริง
*   Keywords พิเศษ:break: สั่งให้หยุดและทะลุออกจาก Loop ทันทีcontinue: สั่งให้ข้ามโค้ดส่วนที่เหลือในรอบนั้น แล้วไปเริ่มต้นรอบถัดไปทันที

**4. Filtering (การคัดกรองข้อมูล)กระบวนการเลือกเฉพาะข้อมูลที่ต้องการ ออกมาจากกลุ่มข้อมูลขนาดใหญ่ โดยเป็นการนำ Logic และ Loop ทำงานร่วมกัน**

*   วิธีการแบบดั้งเดิม (Imperative): ใช้ For Loop วิ่งไล่ดูข้อมูลทุกตัว แล้วใช้ If ตรวจสอบตรรกะ ตัวไหนผ่านเงื่อนไขก็นำมาเก็บไว้ในผลลัพธ์ใหม่
*   วิธีการแบบสมัยใหม่ (Functional): การใช้ฟังก์ชันสำเร็จรูป เช่น ฟังก์ชัน filter() ใน Python/JavaScript หรือการใช้คำสั่ง Where ใน SQL/LINQ เพื่อเขียนโค้ดให้สั้นและอ่านง่ายขึ้น

**Comparison Operators (ตัวดำเนินการเปรียบเทียบ)**

ใช้เปรียบเทียบค่าระหว่างตัวแปร ผลลัพธ์จะเป็น Boolean (True หรือ False)

```python
x = 10
y = 5

print(x > y)
print(x == y)
print(x != y)
```

Boolean Operators ใช้รวมหลายเงื่อนไขเข้าด้วยกัน

```python
age = 20
gpa = 3.5

print(age >= 18 and gpa >= 3.0)
print(age < 18 or gpa >= 3.0)
print(not(age < 18))
```

if-elif-else ใช้ควบคุมการทำงานตามเงื่อนไข

if condition1:
    statement1

elif condition2:
    statement2

else:
    statement3

```python
score = 82

if score >= 80:
    grade = "A"

elif score >= 70:
    grade = "B"

elif score >= 60:
    grade = "C"

else:
    grade = "F"

print(grade)
```

5. Filtering DataFrame ด้วย Logic

```python
import pandas as pd

data = {
    "Name": ["Anna","Ben","Chris","David"],
    "Age": [20,21,22,19],
    "Score": [85,90,75,65]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

กรองข้อมูลด้วยเงื่อนไขเดียว

```python
df[df["Score"] > 80]
```

```python
#การใช้ AND
df[(df["Age"] > 20) &
   (df["Score"] > 80)]
```

```python
#การใช้ OR
df[(df["Age"] < 20) |
   (df["Score"] > 85)]
```

```python
#การใช้ NOT
df[~(df["Score"] > 80)]
```

6. while Loop ใช้ทำงานซ้ำจนกว่าเงื่อนไขจะเป็นเท็จ

```python
#ตัวอย่างทำงานซ้ำจนกว่าเงื่อนไขจะเป็นเท็จ
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
#ตัวอย่างสะสมผลรวม

total = 0
i = 1

while i <= 10:
    total += i
    i += 1

print(total)
```

image.png

```python
#ถ้าต้องการดูการเปลี่ยนแปลงทุกรอบ
total = 0
i = 1

while i <= 10:
    total += i
    print("i =", i, "total =", total)
    i += 1
```

7. for Loop กับ List

```python
fruits = ["Apple",
          "Banana",
          "Orange"]

for fruit in fruits:
    print(fruit)
```

```python
#ใช้ enumerate()

for i, fruit in enumerate(fruits):
    print(i, fruit)
```

8. for Loop กับ Dictionary

```python
student = {
    "Name": "Anna",
    "Age": 20,
    "GPA": 3.8
}
```

```python
#วนเฉพาะ Key

for key in student:
    print(key)
```

```python
#วน Key และ Value

for key, value in student.items():
    print(key, value)
```

9. for Loop กับ NumPy Array

```python
import numpy as np

arr = np.array([10,20,30,40])
```

```python
#วนทีละสมาชิก
for x in arr:
    print(x)
```

```python
#Array 2 มิติ

arr2 = np.array([
    [1,2],
    [3,4]
])

for row in arr2:
    print(row)
```

10. for Loop กับ DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "Name":["Anna","Ben","Chris"],
    "Score":[85,90,78]
})
```

```python
#วนคอลัมน์
for col in df.columns:
    print(col)
```

```python
#วนทีละแถวด้วย iterrows()
for index, row in df.iterrows():

    print(row["Name"],
          row["Score"])
```

11. break และ continue

```python
#break หยุด Loop ทันที
for i in range(10):

    if i == 5:
        break

    print(i)
```

```python
#continue ข้ามรอบปัจจุบัน
for i in range(6):

    if i == 3:
        continue

    print(i)
```
