**Dictionaries & Pandas**

In Python, dictionaries store key-value pairs. Because they organize data in structured mappings, they are frequently used to create and manipulate Pandas DataFrames, which are 2D tabular data structures

```python
import pandas as pd
```

**Dictionary of Lists (Columns)**

The most common way to build a DataFrame is by using a dictionary where keys represent column names and values are lists of data rows.

```python
# 1. Define the dictionary
data_dict = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Paris"]
}

# 2. Convert to Pandas DataFrame
df = pd.DataFrame(data_dict)

print(df)
```

**List of Dictionaries (Rows)**

If your data represents individual records or observations, a list of dictionaries works perfectly, as Pandas will automatically map each dictionary key to a column.

```python
# 1. List of dictionaries
records = [
    {"Name": "Alice", "Age": 25},
    {"Name": "Bob", "Age": 30},
    {"Name": "Charlie"} # Missing 'Age' will be handled gracefully
]

# 2. Convert to Pandas DataFrame
df_records = pd.DataFrame(records)

print(df_records)
```

Structer: pandas.DataFrame(data, index, columns, dtype, copy)

• Data can be ndarray, series, map, list, dict, constant, or another DataFrame.

• Index: row labels; defaults to np.arange(n) if none given.

• Columns: column labels; also default to np.arange(n) if none given.

• dtype: data type of columns.

• copy: copies data if True; default is False.

1. การสร้าง Dictionary

```python
student = {
    "id": 101,
    "name": "Somchai",
    "age": 20,
    "major": "Statistics"
}

print(student)
```

2. การเข้าถึงข้อมูล

```python
#ใช้ Key
print(student["name"])
```

```python
print(student["age"])
```

```python
#ใช้ get()
#ป้องกัน Error หากไม่มี Key

print(student.get("age"))
print(student.get("gpa"))
```

3. การเพิ่มข้อมูล

```python
student["gpa"] = 3.75
print(student)
```

4. การอัปเดตข้อมูล

```python
student["age"] = 21
print(student)
```

5. การลบข้อมูล

```python
if "major" in student:
    del student["major"]
    print("Key 'major' deleted successfully.")
else:
    print("Key 'major' does not exist in the dictionary.")
print(student)
```

6. การวนลูปใน Dictionary

```python
for key in student:
    print(key)
```

```python
for key, value in student.items():
    print(key, value)
```

11. Subset ด้วย loc[]

loc ใช้ ชื่อแถวและชื่อคอลัมน์

จากตัวอย่างเดิม

```python
data = {
    "Name": ["Anna", "Ben", "Chris"],
    "Age": [20, 21, 22],
    "Score": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
#df.loc[rows, columns]

df.loc[1]
```

เลือกหลายแถว

```python
df.loc[0:1]
```

เลือกบางคอลัมน์

```python
df.loc[:, ["Name", "Score"]]
```

เลือกแถวและคอลัมน์พร้อมกัน
