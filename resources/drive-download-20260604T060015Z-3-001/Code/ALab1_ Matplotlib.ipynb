**1. Matplotlib**
Matplotlib is a Python 2D plotting library which produces publication quality figures in a variety of hardcopy formats and interactive environments across platforms. Matplotlib can be used in Python scripts, the Python and IPython shells, the Jupyter notebook, web application servers, and four graphical user interface toolkits.

**Import Package**

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

- The plot() function is used to draw points (markers) in a diagram.

- By default, the plot() function draws a line from point to point.

- The function takes parameters for specifying points in the diagram.

- Parameter 1 is an array containing the points on the x-axis (horizontal axis).

- Parameter 2 is an array containing the points on the y-axis (vertical axis).

**1.1 Line Graph **

```python
#Example1

months = ['Jan','Feb','Mar','Apr','May']
sales = [120,140,180,170,200]

plt.plot(months, sales)
plt.show()
```

**การปรับแต่ง Line Graph **

```python
#Example2

plt.plot(months,
         sales,
         marker='o',
         color='red')

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

plt.grid(True)
plt.show()
```

```python
#Example3 Plot Line

x = np.array([0, 6])
y = np.array([0, 250])

plt.plot(x, y)
plt.show()
```

```python
#Example4 without Line

x = np.array([0, 6])
y = np.array([0, 250])

plt.plot(x, y,'o')
plt.show()
```

```python
#Example5 Multiple Point

x = np.array([1, 2, 6, 8])
y = np.array([3, 8, 1, 10])
plt.plot(x, y)
plt.show()
```

```python
#Example6 Default X-Axis

y = np.array([3,8,1,10,5,7])
plt.plot(y)
plt.show()
```

```python
#Example7 Markers

y = np.array([3,8,1,10])
plt.plot(y, marker='o')
plt.show()
```

**Create Labels & TiTle**

```python
#Example13 Create Labels & TiTle

x = np.array([0,1,2,3,4,5])
y = np.array([0,8,12,20,26,38])

plt.plot(x, y)
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Sport Data')
plt.show()
```

Full Option

```python
#Example14 Set Location Title

x = np.array([0,1,2,3,4,5])
y = np.array([0,8,12,20,26,38])
font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}
plt.plot(x,y)
plt.xlabel('Overs', fontdict=font2)
plt.ylabel('Runs', fontdict=font2)
plt.title('Sport Data', fontdict=font1)
plt.show()

#Set Location
#plt.title('Sport Data', loc='left')
```

**Add Grid**

```python
#Example15 Grid

x = np.array([0,1,2,3,4,5])
y = np.array([0,8,12,20,26,38])
plt.plot(x,y)
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Sport Data', loc='left')
plt.grid()
plt.show()
```

Use subplot() to draw multiple plots in one figure. It takes three arguments: rows, columns, and the plot's position index.

```python
#Example16 subplot()

x = np.array([0,1,2,3])
#plot1
plt.subplot(1,2,1)
plt.plot(x, [3,8,1,10])
#plot2
plt.subplot(1,2,2)
plt.plot(x, [10,20,30,40])
plt.show()
```

```python
#Example17 Multiple Plot

x = np.array([0,1,2,3])

#plot1
plt.subplot(2,1,1)
plt.plot(x, [3,8,1,10])
#plot2
plt.subplot(2,1,2)
plt.plot(x, [10,20,30,40])

plt.show()
```

**1.2 Scatter Plot**

```python
#Example18

height = [150,155,160,165,170,175]
weight = [50,55,58,65,68,72]

plt.scatter(height, weight)

plt.title('Height vs Weight')
plt.xlabel('Height')
plt.ylabel('Weight')

plt.show()
```

```python
#Example19

x = np.array([0,1,2,3,4,5])
y = np.array([0,8,12,20,26,38])

plt.scatter(x, y)
plt.show()
```

**1.3 Bar Chart**

```python
#Example20 Vertical Bar Chart

x = np.array(['A','B','C','D'])
y = np.array([3,8,1,10])
plt.bar(x, y)
plt.show()
```

```python
#Example21 Herizontal Bar Chart

x = np.array(['A','B','C','D'])
y = np.array([3,8,1,10])
plt.barh(x, y)
plt.show()
```

**1.4 Histogram**

```python
#Example22

scores = [60,65,70,72,75,78,80,82,85,88,90,92,95]
plt.hist(scores,bins=5)
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.show()
```

```python
#Example23

x = np.random.normal(170, 10, 250)
#plt.hist(x,bins=5)
plt.hist(x)
plt.show()
```

**1.5 Pie Chart**

```python
#Example24

x = np.array([35,25,25,15])
labels = ["Apples", "Bananas", "Cherries", "Dates"]
plt.pie(x, labels=labels)
#plt.legend(title='Four Fruits:')
plt.show()
```

```python
#Example25 explode parameter

x = np.array([35,25,25,15])
labels = ["Apples", "Bananas", "Cherries", "Dates"]
explode = [0.2, 0, 0, 0]
plt.pie(x, labels=labels, startangle=90, explode=explode)
plt.show()
```

```python
#Example25 colors

x = np.array([35,25,25,15])
labels = ["Apples", "Bananas", "Cherries", "Dates"]
colors = ["black", "hotpink", "b", "#4CAF50"]
explode = [0.8, 0.8, 0.8, 0.8]
plt.pie(x, labels=labels, startangle=90, explode=explode, colors=colors)
plt.show()
```
