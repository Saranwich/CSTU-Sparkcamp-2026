```python
import matplotlib.pyplot as plt 
```

simple plot

```python
x = [1,2,3,4,5]
y = [50,40,30,20,10]
```

```python
plt.plot(x, marker = "v")
plt.show()
```

customization

1. Marker

```python
plt.plot(x, y, marker = "v")
plt.show()


plt.plot(y,x, marker = "o", ms=20)
plt.show()

#Markers Edge
plt.plot(x, y, marker = "x", ms = 10, mec='r')
plt.show()

#Markers Edge
plt.plot(x, y, marker = "x", ms = 20, mfc='r')
plt.show()
```

2. Line

```python
fig, ax = plt.subplots(2,2)
ax[0,0].plot([0,1,2],[0,1,2])
ax[1,1].plot([10,9,8],[100,99,1])
ax[0,1].plot([0,1,-12],[10,12,-82])
ax[1,0].plot([10,9,8],[100,99,31])
plt.show()
```
